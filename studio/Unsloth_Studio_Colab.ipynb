```python
from unsloth import FastLanguageModel
import torch

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="Qwen/Qwen2.5-7B-Instruct",
    max_seq_length=2048,
    dtype=torch.float16,
    load_in_4bit=True,
)

print("Modelo cargado con Unsloth correctamente")
print("GPU:", torch.cuda.get_device_name(0))
print("4 bits:", model.is_loaded_in_4bit)
```

```python
from unsloth import FastLanguageModel

model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    target_modules=[
        "q_proj",
        "k_proj",
        "v_proj",
        "o_proj",
        "gate_proj",
        "up_proj",
        "down_proj",
    ],
    lora_alpha=16,
    lora_dropout=0,
    bias="none",
    use_gradient_checkpointing="unsloth",
    random_state=3407,
    use_rslora=False,
)

model.print_trainable_parameters()
```

```python
from trl import SFTTrainer, SFTConfig

training_args = SFTConfig(
    output_dir="/content/SocialAI_prueba",

    per_device_train_batch_size=1,
    gradient_accumulation_steps=4,

    max_steps=3,
    learning_rate=2e-4,
    warmup_steps=0,

    fp16=True,
    bf16=False,

    optim="adamw_8bit",
    weight_decay=0.01,

    logging_steps=1,
    save_strategy="no",
    report_to="none",

    max_length=512,
    dataset_text_field="text",
    packing=False,
    assistant_only_loss=False,

    gradient_checkpointing=True,
    seed=3407,
)

trainer = SFTTrainer(
    model=model,
    args=training_args,
    train_dataset=dataset,
    processing_class=tokenizer,
)

print("Entrenador configurado correctamente.")
```

```python
resultado = trainer.train()

print("\nEntrenamiento finalizado.")
print("Loss final:", resultado.training_loss)
print("Pasos realizados:", resultado.global_step)
```

```python
import gradio as gr
import torch

FastLanguageModel.for_inference(model)

def analizar_conversacion(conversacion):
    if not conversacion.strip():
        return "Pega una conversación primero.", "", "", ""

    instrucciones = """
Eres SocialAI, un copiloto de conversaciones reales.
Analiza conversaciones y ayuda a responder de forma natural,
respetuosa y auténtica.

No inventes sentimientos ni intenciones de las personas.
No recomiendes manipular, presionar ni insistir.

Devuelve exactamente este formato:

ANALISIS:
[Explica brevemente el contexto.]

RESPUESTA 1 - NATURAL:
[Una respuesta natural.]

RESPUESTA 2 - DIVERTIDA:
[Una respuesta ligera y respetuosa.]

RESPUESTA 3 - TRANQUILA:
[Una respuesta sencilla y sin presión.]
- Detecta respuestas cortas o monosilábicas como:
  "sí", "no", "bien", "ok", "jaja", "normal" y "no sé".

- Si la otra persona responde brevemente:
  1. No repitas la misma pregunta.
  2. No fuerces la conversación.
  3. Busca una forma natural de cambiar de tema.
  4. Aprovecha los detalles de mensajes anteriores.
  5. Si varias respuestas consecutivas son cortas,
     recomienda dar espacio y no insistir.

- No confundas una respuesta corta con falta de interés.
  No puedes conocer las intenciones de la otra persona.

- La respuesta divertida debe tener humor natural
  y nunca parecer una frase sacada de un catálogo.

- La respuesta tranquila debe ser sencilla,
  auténtica y sin presión.

- Cada alternativa debe cumplir una función diferente.
"""

    mensajes = [
        {"role": "system", "content": instrucciones},
        {
            "role": "user",
            "content": f"Analiza esta conversación:\n{conversacion}"
        }
    ]

    entradas = tokenizer.apply_chat_template(
        mensajes,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt"
    ).to("cuda")

    with torch.inference_mode():
        salida = model.generate(
            input_ids=entradas,
            max_new_tokens=350,
            do_sample=True,
            temperature=0.7,
            top_p=0.9,
            use_cache=True,
        )

    texto = tokenizer.decode(
        salida[0][entradas.shape[-1]:],
        skip_special_tokens=True
    )

    # Separar las cuatro secciones
    import re

    secciones = re.split(
        r"ANALISIS:|RESPUESTA 1 - NATURAL:|RESPUESTA 2 - DIVERTIDA:|RESPUESTA 3 - TRANQUILA:",
        texto,
        flags=re.IGNORECASE
    )

    secciones = [s.strip() for s in secciones if s.strip()]

    while len(secciones) < 4:
        secciones.append("No se pudo generar esta sección. Inténtalo nuevamente.")

    return tuple(secciones[:4])


with gr.Blocks(theme=gr.themes.Soft(), title="SocialAI") as demo:

    gr.Markdown("# 💬 SocialAI")
    gr.Markdown("### Tu copiloto de conversaciones reales")
    gr.Markdown(
        "Pega una conversación y recibe sugerencias para continuar "
        "de forma natural y respetuosa."
    )

    conversacion = gr.Textbox(
        label="Tu conversación",
        placeholder="Yo: Hola, ¿cómo estuvo tu día?\nElla: Bien jajaja...",
        lines=8
    )

    boton = gr.Button("✨ Analizar conversación", variant="primary")

    analisis = gr.Textbox(label="🔎 Análisis del contexto", lines=3)

    with gr.Row():
        natural = gr.Textbox(label="😊 Respuesta natural", lines=4)
        divertida = gr.Textbox(label="😄 Respuesta divertida", lines=4)
        tranquila = gr.Textbox(label="🙂 Respuesta tranquila", lines=4)

    boton.click(
        fn=analizar_conversacion,
        inputs=conversacion,
        outputs=[analisis, natural, divertida, tranquila]
    )

demo.launch(share=True, debug=True)
```
